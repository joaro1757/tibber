```python
from pyspark.sql import functions as F

landing_path = "/Volumes/workspace/landing"
sources = ["meter_readings", "spot_prices", "homes", "subscriptions"]

spark.sql("create schema if not exists workspace.bronze")

for source in sources:
    (
        spark.read.csv(f"{landing_path}/{source}", header=True)
        .withColumns({
            "_source_file": F.col("_metadata.file_path"),
            "_loaded_at": F.current_timestamp(),
        })
        .write.mode("overwrite")
        .saveAsTable(f"workspace.bronze.{source}")
    )
```
