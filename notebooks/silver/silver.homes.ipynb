```python
%sql

create or replace table workspace.silver.homes
comment '''
    one row per home and price area validity period. a home can change price area, so it can have multiple price areas within one month. periods are half-open: valid_from <= date < valid_to. null valid_to means current.
'''
as

select
    trim(h.home_id) as home_id,
    upper(trim(h.country)) as country,
    upper(trim(h.price_area)) as price_area,
    cast(h.valid_from as date) as valid_from,
    cast(h.valid_to as date) as valid_to,
    current_timestamp() as _processed_at
from 
    workspace.bronze.homes h
```
